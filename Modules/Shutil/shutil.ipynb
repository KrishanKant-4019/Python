# 1. Importing the Module

```python
import shutil
import os
```

# 2. Copying Files and Directories

#### Copy a single file

```python
shutil.copy("source.txt", "destination.txt")
```

#### Copy with metadata

```python
shutil.copy2("source.txt", "destination_with_metadata.txt")
```

#### Copy an entire directory

```python
shutil.copytree("source_folder", "destination_folder")

# If the destination already exists, it’ll throw an error — so use:
# shutil.copytree("source_folder", "destination_folder", dirs_exist_ok = True)
```

# 3. Moving and Renaming

#### Move file or folder

```python
shutil.move("source_folder/sample.txt", "destination_folder/sample.txt")
```

# 4. Deleting Files and Directories

#### Delete a directory tree

```python
# shutil.rmtree("source_folder")
```

#### Delete single file (via os)

```python
os.remove("destination_with_metadata.txt")
```

# 5. Archiving (Compressing/Extracting)

#### Create a ZIP archive

```python
shutil.make_archive("back", "zip", "destination_folder")
```

#### Extract an archive

```python
shutil.unpack_archive("back.zip", "extracted_folder")
```

# 6. Disk Usage Information

```python
usage = shutil.disk_usage("/")
print(f"Total: {usage.total // (2**30)} GB")
print(f"Used: {usage.used // (2**30)} GB")
print(f"Free: {usage.free // (2**30)} GB")
```

# 7. Copying File Objects (Streams)

```python
with open("source.txt", "rb") as src, open("destination.txt", "wb") as dest:
    shutil.copyfileobj(src, dest)
```

# 8. Ignore Certain Files When Copying a Directory

```python
def ignore_pyc(dir, files):
    return [f for f in files if f.endswith(".pyc") or f == "__pycache__"]
shutil.copytree("source_folder", "destination_folder_2", ignore = ignore_pyc)    
```
