```python
import calendar
print(calendar.month(2025,5)) #Returns a calendar for a given month.
```

```python
print(calendar.calendar(2025)) #Returns a calendar for a given year.
```

```python
print(calendar.isleap(2025)) #Check a year is leap or not.
```

```python
print(calendar.leapdays(2000,2025)) #Returns total number of leap years between two given year.
```

```python
print(calendar.weekday(2025,5,20)) #Returns the weekday of a given date as an integer:
```

```python
print(calendar.monthrange(2025,5)) #Returns a tuple: (start_day, number_of_days)
```
