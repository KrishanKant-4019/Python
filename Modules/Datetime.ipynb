# 1. Importing and Basic Structure

```python
import datetime

print(datetime.datetime.now())
```

# 2. Get Current Date and Time

```python
from datetime import datetime

now = datetime.now()
print("Current Date & Time:", now)
print("Date:", now.date())
print("Time:", now.time())
```

# 3. Get Today’s Date Only

```python
from datetime import date

today = date.today()
print("Today's Date:", today)
print("Year:", today.year)
print("Month:", today.month)
print("Day:", today.day)
```

# 4. Create a Specific Date or Time

```python
from datetime import date, time, datetime

d = date(2025, 5, 17)
t = time(14, 42, 0)
dt = datetime(2025, 5, 17, 14, 42, 0)

print("Date object:", d)
print("Time object:", t)
print("Datetime object:", dt)
```

# 5. Formatting Dates

```python
from datetime import datetime

now = datetime.now()
print(now.strftime("%Y-%m-%d"))
print(now.strftime("%B %d %Y"))
print(now.strftime("%I:%M %p"))
print(now.strftime("%A"))
```

# 6. Parsing Strings into Dates

```python
from datetime import datetime

date_str = "07-11-2025 14:30"
converted = datetime.strptime(date_str, "%d-%m-%Y %H:%M")

print("Converted:", converted)
print("Year:", converted.year)
print("Month:", converted.month)
```

# 7. Date Arithmetic

```python
from datetime import datetime, timedelta

today = datetime.now()
yesterday = today - timedelta(days = 1)
next_week = today + timedelta(weeks = 1)

print("Today:", today)
print("Yesterday:", yesterday)
print("Next Week:", next_week)
```

# 8. Difference Between Two Dates

```python
from datetime import date

d1 = date(2025, 11, 7)
d2 = date(2025, 12, 25)

diff = d2 - d1
print("Difference:", diff.days)
```

# 9. Working with Timezones

```python
from datetime import datetime
import pytz

utc_time = datetime.now(pytz.utc)
india_time = utc_time.astimezone(pytz.timezone("Asia/Kolkata"))
us_time = utc_time.astimezone(pytz.timezone("US/Eastern"))

print("UTC Time:", utc_time)
print("India Time:", india_time)
print("US Time:", us_time)
```

# 10. Practical Example — Countdown Timer

```python
from datetime import datetime, timedelta
import time

end_time = datetime.now() + timedelta(seconds = 5)
while datetime.now() < end_time:
    remaining = end_time - datetime.now()
    print(f"Time left: {remaining.seconds} sec", end = "\r")
    time.sleep(1)

print("\nTime's up!")
```
