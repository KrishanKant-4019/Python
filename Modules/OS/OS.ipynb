# 6. File and Directory Information

#### Get File Size

```python
print(os.path.getsize("OS/OS.ipynb")) # in bytes
```

#### Get File Creation / Modification Time

```python
import datetime

path = "OS/OS.ipynb"
ctime = os.path.getctime(path)
mtime = os.path.getmtime(path)

print("Creation Time:", datetime.datetime.fromtimestamp(ctime))
print("Modification Time:", datetime.datetime.fromtimestamp(mtime))
```

# 7. Walking Through Directories

```python
for root, dirs, files in os.walk("C:\\Users\\krish\\Documents"):
    print("Current Path:", root)
    print("Directories:", dirs)
    print("Files:", files)
```
