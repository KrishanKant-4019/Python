```python
import os

folder_path = r"C:\Users\krish\Python\Modules\OS\Mini_Folder_Organizer"

os.chdir(folder_path)
print("Current Directory:", os.getcwd())

categories = {
    "Images": [".jpg", ".jpeg", ".png", ".gif"],
    "Documents": [".pdf", ".docx", ".txt"],
    "Videos": [".mp4", ".mkv", ".avi"],
    "Data": [".csv", ".xlsx", ".json"]
}

# Create folders if they don’t exist
for folder in categories.keys():
    if not os.path.exists(folder):
        os.mkdir(folder)
print("Folder created successfully!")

for file in os.listdir():
    if os.path.isfile(file):
        name, ext = os.path.splitext(file)

        if ext == "":
            continue

        moved = False
        for folder, extensions in categories.items():
            if ext.lower() in extensions:
                new_path = os.path.join(folder, file)
                os.rename(file, new_path)
                print(f"Moved {file} -> {folder}")
                moved = True
                break

        if not moved:
            other_folder = "Others"
            if not os.path.exists(other_folder):
                os.mkdir(other_folder)
            new_path = os.path.join(other_folder, file)
            os.rename(file, new_path)
            print(f"Moved {file} -> {other_folder}")

with open("organizer_log.txt", "w") as log:
    log.write("File organized successfully in:\n")
    log.write(folder_path + "\n")
    log.write("Subfolders created:\n")
    for folder in categories.keys():
        log.write(f"-{folder}\n")

print("\nOrganization complete!")
```
