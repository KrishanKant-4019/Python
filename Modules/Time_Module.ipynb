```python
import time

#Returns current time in seconds since epoch
current_seconds=time.time() 
print("Seconds since epoch:",current_seconds)

#Converts seconds since epoch to a readable string
print("Current time:",time.ctime(current_seconds)) 

#Pauses execution for specified seconds
print("Pausing for 3 seconds...")
time.sleep(3) 
print("Resumed at:",time.ctime())

# Format current local time
formatted_time = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime())
print("Formatted local time:", formatted_time)
```

```python
# Get current local time as a struct_time object
local_time = time.localtime()
print("Local time struct:", local_time)

# Accessing components
print("Year:", local_time.tm_year)
print("Month:", local_time.tm_mon)
print("Day:", local_time.tm_mday)
```

```python
# Get current local time
local_time = time.localtime()

# Format as "YYYY-MM-DD HH:MM:SS"
formatted = time.strftime("%Y-%m-%d %H:%M:%S", local_time)
print("Formatted local time:", formatted)
```

```python
# Parse a date string into a struct_time object
date_string = "20 Apr, 2025 13:39:00"
parsed_time = time.strptime(date_string, "%d %b, %Y %H:%M:%S")
print("Parsed struct_time:", parsed_time)
```
