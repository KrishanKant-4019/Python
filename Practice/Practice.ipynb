```python
# Write a Python program to find the least common multiple (LCM) of two positive integers(By Euclidean's Algorithm).
a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))

# Finding GCD
x, y = a, b
while y != 0:
    x, y = y, x%y

gcd = x

# Finding LCM
lcm = abs(a*b)//gcd
print("LCM is:", lcm)
```

```python
# Write a Python program to convert temperatures to and from Celsius and Fahrenheit.
# [ Formula : c/5 = f-32/9 [ where c = temperature in celsius and f = temperature in fahrenheit ]
celsius = float(input("Enter the temperature in celsius:"))
fahrenheit = float(input("Enter the temperature in fahrenheit:"))

C_to_F = (9.0/5.0)*celsius + 32
F_to_C = (fahrenheit - 32) * 5.0 / 9.0

print(f"{celsius} degree Celsius is equal to {C_to_F} degree Fahrenheit.")
print(f"{fahrenheit} degree Fahrenheit is equal to {F_to_C} degree Celsius.")
```

```python
# Write a Python function to calculate the factorial of a number (a non-negative integer). 
# The function accepts the number as an argument.
def fact(n):
    result = 1
    while n > 1:
        result *= n
        n -= 1
    return result
    
print(fact(5))
```

```python
# Count Uppercase and Lowercase Letters in a String
myString = "Krishan Kant Jha"
def count_case(myString):
   upper_count = 1
   lower_count = 1
   for char in myString:
       if char.isupper():
            upper_count += 1
       elif char.islower():
            lower_count += 1
   print("Uppercase Letters:", upper_count)
   print("Lowercase Letters:", lower_count)

count_case(myString)
```
