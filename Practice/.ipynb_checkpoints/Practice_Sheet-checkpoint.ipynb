```python
# Even or Odd from 1 to N
def even_or_odd(n):

    for i in range(1, n+1):
        if i % 2 == 0:
            print(i, "is Even")

        else:
            print(i, "is Odd")

num = int(input("Enter the number:"))
even_or_odd(num)
```

```python
# Sum of Digits of a Number
def sum_of_digits(n):
    total = 0

    while n > 0:
        digit = n % 10
        total += digit
        n = n // 10

    return total

num = int(input("Enter the number:"))
print("Sum of digits:", sum_of_digits(num))
```

```python
# Reverse the Digits of a Number
def reverse_digit(n):
    rev = 0
    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10
    return rev

num = int(input("Enter the number:"))
print(reverse_digit(num))
```

```python
# Check if Number is Palindrome
def is_palindrome(n):
    original = n
    rev = 0

    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10

    if rev == original:
        return True
    else:
        return False

num = int(input("Enter the number:"))
print(is_palindrome(num))
```

```python
# Check if Number is Prime
def is_prime(n):
    if n <= 1:
        return False
        
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

num = int(input("Enter the number:"))
print(is_prime(num))
```

```python
# Print All Prime Numbers from 1 to N
def all_prime(x):
    for i in range(1, x+1):
        if is_prime(i):
            print(i, end = " ")

num = int(input("Enter the number:"))
all_prime(num)
```

```python
# Count Prime Numbers from 1 to N
def count_prime(x):
    count = 0
    for i in range(1, x+1):
        if is_prime(i):
            count += 1
    return count

num = int(input("Enter the number:"))
print("Total prime numbers:", count_prime(num))
```

```python
# Find Factorial of a Number
def factorial(n):
    if n < 0:
        return -1
    elif n == 0 or n == 1:
        return 1
    else:
        fact = 1
        for i in range(1, n+1):
            fact *= i
        return fact

num = int(input("Enter the number:"))
print("Factorial is:", factorial(num))
```

```python
# Check if Number is Armstrong
def is_armstrong(n):
    original = n
    length = len(str(n))
    result = 0

    while n > 0:
        digit = n % 10
        result += pow(digit, length)
        n //= 10

    if result == original:
        return True
    else: 
        return False

num = int(input("Enter the number:"))
print(is_armstrong(num))
```

```python
# Check if Number is Perfect
def is_perfect(n):
    original = n
    result = 0

    for i in range(1, n):
        if n % i == 0:
            result += i
            
    if result == original:
        return True
    else:
        return False

num = int(input("Enter the number:"))
print(is_perfect(num))
```

```python
# Product of Digits
def product_digits(n):
    total = 1

    while n > 0:
        digit = n % 10
        total *= digit
        n //= 10
    return total

num = int(input("Enter the number:"))
print(product_digits(num))
```

```python
# Find Largest Digit in a Number
def largest_digit(n):
    n = abs(n)
    max_digit = 0

    while n > 0:
        digit = n % 10
        if digit > max_digit:
            max_digit = digit
        n //= 10
    return max_digit

num = int(input("Enter the number:"))
print("Largest digit:", largest_digit(num))
```

```python
# Find Smallest Digit in a Number
def smallest_digit(n):
    n = abs(n)
    min_digit = 9

    while n > 0:
        digit = n % 10
        if digit < min_digit:
            min_digit = digit

        n //= 10
    return min_digit

num = int(input("Enter the number:"))
print("Smallest digit:", smallest_digit(num))
```

```python
# Check if All Digits are Even
def all_even(n):
    n = abs(n)

    if n == 0:
        return True
        
    while n > 0:
        digit = n % 10
        if digit % 2 != 0:
            return False
        n //= 10
    return True

num = int(input("Enter the number:"))
print("Even digits:", all_even(num))
```

```python
# Count Even and Odd Digits in a Number
def even_odd(n):
    n = abs(n)
    even_count = 0
    odd_count = 0

    while n > 0:
        digit = n % 10
        if digit % 2 == 0:
            even_count += 1
        else:
            odd_count += 1
        n //= 10   
        
    return even_count, odd_count

num = int(input("Enter the number:"))
even_count, odd_count = even_odd(num)
print("Even Count:", even_count)
print("Odd Count:", odd_count)
```

```python
# Find Sum of Even Digits and Odd Digits Separately
def even_odd_sum(n):
    n = abs(n)
    even_sum = 0 
    odd_sum = 0

    while n > 0:
        digit = n % 10
        if digit % 2 == 0:
            even_sum += digit
        else:
            odd_sum += digit
        n //= 10
    return even_sum, odd_sum

num = int(input("Enter the number:"))
even_sum, odd_sum = even_odd_sum(num)
print("Even Sum:", even_sum)
print("Odd Sum:", odd_sum)
```

```python
# Check if Number has Duplicate Digits
def has_duplicate_digits(number):
    num_str = str(abs(number))    
    if len(num_str) != len(set(num_str)):
        return True
    return False
    
num = int(input("Enter the number"))
print(has_duplicate_digits(num))  
```
