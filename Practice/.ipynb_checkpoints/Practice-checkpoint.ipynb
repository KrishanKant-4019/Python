```python
# Write a Python program that accepts a sequence of comma-separated numbers from the user and generates 
# a list and a tuple of those numbers.
num=str(input("Enter the numbers:"))
num_list=num.split(",")
num_tuple=tuple(num_list)

print("List:",num_list)
print("Tuple:",num_tuple)
```

```python
# Write a Python program to display the first and last colors from the following list.
# color_list = ["Red","Green","White" ,"Black"]
from operator import itemgetter

color_list=["Red","Green","White","Black"]
getter=itemgetter(0,3)
print(getter(color_list))
```

```python
# Write a Python program to get the volume of a sphere with radius six.
import math

radius=6
volume=(4/3)*math.pi*radius*radius*radius
print("The volume of sphere is:",volume)
```

```python
# Write a Python program to test whether a number is within 100 of 1000 or 2000.
num=int(input("Enter the number:"))

if abs(1000-num)<=100 or abs(2000-num)<=100:
    print("The number is within 100 of 1000 or 2000")
else:
    print("The number is not within 100 of 1000 or 2000")
```

```python
# Write a Python program to calculate the sum of three given numbers. 
# If the values are equal, return three times their sum.
a=int(input("Enter the 1st number:"))
b=int(input("Enter the 2nd number:"))
c=int(input("Enter the 3rd number:"))

if a==b and b==c:
    print(3*(a+b+c))
else:
    print(a+b+c)
```

```python
# Write a Python program to calculate the difference between a given number and 17. 
# If the number is greater than 17,return twice the absolute difference.
num=int(input("Enter the number"))

if num>17:
    print(2*abs(num-17))
else:
    print(abs(num-17))
```

```python
# Write a Python program that accepts an integer (n) and computes the value of n+nn+nnn
n=input("Enter the number:")
nn = n+n
nnn = n+n+n
result = int(n) + int(nn) + int(nnn)
print("Result:",result)
```

```python
# Write a Python program to calculate the number of days between two dates.
import datetime

x=datetime.datetime(2025,5,20)
print(x)

y=datetime.datetime(2005,7,23)
print(y)

difference = x- y
print("Number of days between dates:", difference.days)
```
