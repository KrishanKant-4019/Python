# Encapsulation

```python
# It is the process of wrapping data and methods into a single unit (class) and restricting direct access.
```

```python
# Example-1 -> Create a class Person with a private attribute __name. Add a method set_name() and get_name() 
# to modify and retrieve it.

class Person:
    def __init__(self, name):
        self.__name = name # Private Variable

    def set_name(self, new_name):
        self.__name = new_name # Set or update the name

    def get_name(self):
        return self.__name

person = Person("Krishan Kant Jha")
print("Initial Name:", person.get_name())

person.set_name("John")
print("Updated Name:", person.get_name())
```

```python
# Example-2 ->  Create a BankAccount Class
# Private attribute: __balance
# Method to deposit(amount)
# Method to withdraw(amount)
# Method to get_balance()

class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Amount Deposited. New Balance : {self.__balance}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Amount Withdrawn. New Balance : {self.__balance}")
        else:
            print("Insufficient funds or invalid amount.")

    def get_balance(self):
        return f"Total Balance : {self.__balance}"

account = BankAccount(50000)
account.deposit(10000)
account.withdraw(5000)
print(account.get_balance())
```

```python
# Example-3 -> Student Marks System
# Private attribute: __marks
# Method to set and get marks
# Validate that marks must be between 0 and 100 in the setter

class Student:
    def __init__(self, marks=0):
        self.__marks = 0
        self.set_marks(marks)

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
            print(f"Marks set to : {self.__marks}")
        else:
            print("Invalid Marks!")

    def get_marks(self):
        return f"Student Marks : {self.__marks}"


student = Student()
student.set_marks(75)
print(student.get_marks())
```

```python
# Example-4 -> # Create class Employee
# Private attribute: __salary
# Method to increase salary with a percentage
# Prevent setting salary below minimum wage (e.g., 10000)

class Employee:
    def __init__(self, salary):
        if salary >= 10000:
            self.__salary = salary
        else:
            self.__salary = 10000
            print(f"Salary set to minimum wage (10000).")
                
    def salary_increment(self, percentage):
        increment = self.__salary * (percentage / 100)
        self.__salary += increment
        print(f"Salary Incremented. New Salary : {self.__salary}")

    def setting_salary(self, wage):
        if wage >= 10000:
            self.__salary = wage
            print(f"Salary updated to : {self.__salary}")
        else:
            print("Salary cannot be set below minimum wage (10000).")

    def get_salary(self):
        return f"Current Salary: {self.__salary}"

emp = Employee(9000)
emp.setting_salary(12000)
emp.salary_increment(10)
print(emp.get_salary())         
```

```python
# Example-5 -> Login System with Private Password
# Class: User
# Private attribute: __password
# Method to set and check password
# Password must be at least 8 characters and contain numbers

class User:
    def __init__(self, password):
        if len(password) >= 8 and any(char.isdigit() for char in password):
            self.__password = password
        else:
            self.__password = ""
            print("Initial password is invalid. Must be at least 8 characters and contain a number.")

    def set_password(self, new_password):
        if len(new_password) >= 8 and any(char.isdigit() for char in new_password):
            self.__password = new_password
        else:
            print("Password must be at least 8 characters and contain a number.")

    def check_password(self, check):
        if self.__password == check:
            print("Password is correct!")
        else:
            print("Password is incorrect!")

user = User("ABCDE123")
user.set_password("NewPass123")
user.check_password("NewPass123")
```
