```python
# Create a Set
thisset={"apple", "banana", "cherry"}
print(thisset)
```

```python
# Duplicates Not Allowed
thisset = {"apple", "banana", "cherry", "apple"}
print(thisset)
```

```python
# True and 1 is considered as same value 
thisset = {"apple", "banana", "cherry", True, 1, 2}
print(thisset)
```

```python
# False and 0 is considered the same value
thisset={"apple", "banana", "cherry", False, True, 0}
print(thisset)
```

```python
# Length of a Set
thisset = {"apple", "banana", "cherry"}
print(len(thisset))
```

```python
#A set can contain different data types
set1 = {"abc", 34, True, 40, "male"}
print(set1)
```

```python
#type()
thisset = {"apple", "banana", "cherry"}
print(type(thisset))
```

```python
# The set() Constructor
thisset=set(("apple","banana","cherry"))
print(thisset)
```

```python
# Access Items
thisset = {"apple", "banana", "cherry"}
for x in thisset:
  print(x)
```

```python
# Add Items
thisset = {"apple", "banana", "cherry"}
thisset.add("orange")
print(thisset)
```

```python
# Add Sets
thisset = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}
thisset.update(tropical)
print(thisset)
```

```python
# Add Any Iterable
thisset = {"apple", "banana", "cherry"}
mylist = ["kiwi", "orange"]
thisset.update(mylist)
print(thisset)
```

```python
# Remove Item(Note: If the item to remove does not exist, remove() will raise an error.
thisset = {"apple", "banana", "cherry"}
thisset.remove("banana")
print(thisset)
```

```python
# Remove Item by discard() method(Note: If the item to remove does not exist, discard() will NOT raise an error.)
thisset = {"apple", "banana", "cherry"}
thisset.discard("banana")
print(thisset)
```

```python
# Remove a random item by using the pop() method:
thisset = {"apple", "banana", "cherry"}
x = thisset.pop()
print(x)
print(thisset)
```

```python
# clear() method empties the set:
thisset = {"apple", "banana", "cherry"}
thisset.clear()
print(thisset)
```

```python
# The del keyword will delete the set completely:
thisset = {"apple", "banana", "cherry"}
del thisset
```

```python
# The union() method returns a new set with all items from both sets.
set1={"a","b","c"}
set2={1,2,3}
set3=set1.union(set2)
print(set3)
```

```python
#Use | to join two sets:
set1={"a","b","c"}
set2={1,2,3}
set3=set1|set2
print(set3)
```

```python
#Join Multiple Sets
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

myset = set1.union(set2, set3, set4)
print(myset)
```

```python
#Join a Set and a Tuple
x = {"a", "b", "c"}
y = (1, 2, 3)

z = x.union(y)
print(z)
```

```python
# .update()(It change's the original set.)
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}

set1.update(set2)
print(set1)
```

```python
# .intersection()(It return's a new set)
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.intersection(set2)
print(set3)
```

```python
#Use & to join two sets:(It allows joining only with another set.)
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1&set2
print(set3)
```

```python
# The intersection_update() method will also keep ONLY the duplicates, but it will change the original 
# set instead of returning a new set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.intersection_update(set2)

print(set1)
```

```python
#The values True and 1 are considered the same value. The same goes for False and 0.
set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

set3 = set1.intersection(set2)

print(set3)
```

```python
#Difference
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.difference(set2)

print(set3)
```

```python
#Use - to join two sets:(It allows joining only with another set.)
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 - set2
print(set3)
```

```python
#The difference_update() method will also keep the items from the first set that are not in the other set, 
# but it will change the original set instead of returning a new set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.difference_update(set2)

print(set1)
```

```python
# The symmetric_difference() method will keep only the elements that are NOT present in both sets.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.symmetric_difference(set2)

print(set3)
```

```python
# Use ^ to join two sets:
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 ^ set2
print(set3)
```

```python
#The symmetric_difference_update() method will also keep all but the duplicates,
# but it will change the original set instead of returning a new set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set1.symmetric_difference_update(set2)
print(set1)
```
