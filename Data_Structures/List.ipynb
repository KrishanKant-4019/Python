```python
#List is changeable and ordered data type in python.
mylist=["apple","banana","mango"]
print(mylist)
```

```python
#List allows duplicate values.
mylist=["apple","apple","banana","mango","banana"]
print(mylist)
```

```python
#Length of the list
mylist=["apple","apple","banana","mango","banana"]
print(len(mylist))
```

```python
#list can be of any data types
list1 = ["apple", "banana", "cherry"]
list2 = [1, 5, 7, 9, 3]
list3 = [True, False, False]

print(list1, list2, list3)
```

```python
#List can contain different data types
mylist = ["apple", True, 15, 76.6]
print(mylist)
```

```python
#Class of the list
mylist=["apple","banana","cherry"]
print(type(mylist))
```

```python
#List() Constructor
mylist=list(("apple","mango","banana"))
print(mylist)
```

```python
#Accesing list items
mylist=list(("apple","mango","banana"))
print(mylist[0])
print(mylist[1])
print(mylist[2])
print(mylist[-1])
print(mylist[0:2])
print(mylist[:3])
print(mylist[1:])
print(mylist[-4:-1])
```

```python
#Check if item exists
mylist=["apple","mango","banana"]
if "apple" in mylist:
    print("apple is present")
```

```python
#Change Item Value
mylist=["apple","mango","banana"]
mylist[1]="cherry"
print(mylist)
```

```python
#Change a Range of Item Values
mylist=["apple", "banana", "cherry", "orange", "kiwi", "mango"]
mylist[1:4]=["watermelon","peach","pomegranate"]
print(mylist)
```

```python
#Insert Items using .insert()
mylist=["apple", "banana", "cherry", "orange", "kiwi", "mango"]
mylist.insert(2,"watermelon")
print(mylist)
```

```python
#Append Items
mylist=list(("Apple","Mango","Banana"))
mylist.append("Cherry")
print(mylist)
```

```python
#Extend List
mylist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]
mylist.extend(tropical)
print(mylist)
```

```python
#Add Any Iterable
mylist=["Apple","Banana","Cherry"]
tuple=("Pineapple","Papaya")
mylist.extend(tuple)
print(mylist)
```

```python
#Remove Specified Item
mylist=["Apple","Banana","Papaya"]
mylist.remove("Papaya")
print(mylist)
```

```python
#Remove Specified Index
mylist=["Apple","Banana","Papaya"]
mylist.pop(1)
print(mylist)
```

```python
#Clear the List
mylist=["Apple","Banana","Papaya"]
mylist.clear()
print(mylist)
```

```python
#Loop Through a List
mylist=["apple", "banana", "cherry", "orange", "kiwi", "mango"]
for x in mylist:
    print(x)
print(mylist[1])
```

```python
#Loop Through the Index Numbers
mylist=["apple", "banana", "cherry", "orange", "kiwi", "mango"]
for i in range(len(mylist)):
  print(mylist[i])
print(mylist[1])
```

```python
mylist = ["apple", "banana", "cherry", "orange", "kiwi", "mango"]
i = 0
while i < len(mylist):
    print(mylist[i])  
    i = i + 1
```

```python
#Looping Using List Comprehension
mylist=["apple","banana","cherry"]
result=[x for x in mylist]
print(result)
```

```python
#Without list Comprehension
fruits=["apple", "banana", "cherry", "kiwi", "mango"]
newlist=[]

for x in fruits:
    if "a" in x:
        newlist.append(x)
print(newlist)
```

```python
#List Comprehension
fruits=["apple", "banana", "cherry", "kiwi", "mango"]
newlist=[x for x in fruits if "a" in x]
print(newlist)
```

```python
#Create a list of squares of numbers from 1 to 10.
numbers=[x**2 for x in range(1,11)]
print(numbers)
```

```python
#Create a list of even numbers from 1 to 20.
even_nums=[x for x in range(1,21) if x%2==0]
print(even_nums)
```

```python
#Convert a list of strings to uppercase.
strings=["Krishan","Hemant","Divya"]
upper_case=[x.upper() for x in strings]
print(upper_case)
```

```python
#Extract digits from a string.
s="Krishan4019"
result=[x for x in s if x.isdigit()]
print(result)
```

```python
#Create a list of characters from a string.
s="Krishan Kant Jha"
character=[x for x in s]
# character=list(s)
print(character)
```

```python
#Get a list of all vowels in a string.
s="Krishan"
vowels=[x for x in s if x in "aeiouAEIOU"]
print(vowels)
```

```python
#Flatten a 2D list.
list=[[1,2],[3,4]]
flattened_list=[item for sublist in list for item in sublist]
print(flattened_list)
```

```python
#Sort List Alphanumerically
mylist=["Apple","Orange","Cherry","Pineapple","Banana"]
mylist.sort()
print(mylist)
```

```python
#Sort Descending
mylist=["Apple","Orange","Cherry","Pineapple","Banana"]
mylist.sort(reverse=True)
print(mylist)
```

```python
#Customize Sort Function
def myfunc(n):
    return abs(n-50)

mylist=[100, 50, 65, 82, 23]
mylist.sort(key=myfunc)
print(mylist)
```

```python
#Case Insensitive Sort
mylist=["banana", "Orange", "Kiwi", "cherry"]
mylist.sort()
print(mylist)
```

```python
#Reverse Order
mylist=["Apple","Orange","Cherry","Pineapple","Banana"]
mylist.reverse()
print(mylist)
```

```python
#Copy a List
mylist=["apple", "banana", "cherry"]
mylist.copy()
print(mylist)
```

```python
#Another way to make a copy is to use the built-in method list().
mylist = ["apple", "banana", "cherry"]
print(list(mylist))
```

```python
#Use the slice Operator
mylist=["apple","banana","cherry"]
print(mylist[:])
```

```python
#Join Two Lists
list1=["apple", "banana", "cherry"]
list2=["pineapple","orange","mango"]
original_list=list1+list2
print(original_list)
```

```python
#Append list2 into list1
list1=["apple", "banana", "cherry"]
list2=["pineapple","orange","mango"]

for x in list2:
    list1.append(x)
print(list1)
```
