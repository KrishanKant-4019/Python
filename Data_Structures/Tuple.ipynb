```python
#Create a Tuple
mytuple=("apple","banana","mango","cherry","pineapple")
print(mytuple)
```

```python
#Tuples allow duplicate values:
mytuple=("apple","banana","banana","apple","pineapple")
print(mytuple)
```

```python
# Tuple Length
mytuple=("apple","banana","mango","cherry","pineapple")
print(len(mytuple))
```

```python
# Create Tuple With One Item
mytuple=("apple",)
print(mytuple)
print(type(mytuple))
```

```python
# A tuple can contain different data types:
mytuple = ("abc", 34, True, 40, "male")
print(mytuple)
```

```python
#The tuple() Constructor
mytuple=tuple(("apple","banana","cherry","orange"))
print(mytuple)
```

```python
# Access Tuple Items
thistuple = ("apple", "banana", "cherry")
print(thistuple[1])
```

```python
# Negative Indexing
thistuple = ("apple", "banana", "cherry")
print(thistuple[-1])
```

```python
# Range of Indexes
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[2:5])
```

```python
# Range of Negative Indexes
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[-4:-1])
```

```python
# Check if Item Exists
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
if "cherry" in thistuple:
    print("Yes, it is present!")
else:
    print("No, it is not present!")
```

```python
#Change Tuple Values
x= ("apple", "banana", "cherry")
y=list(x)
y[1]="mango"
x=tuple(y)
print(x)
```

```python
#Add Items
x= ("apple", "banana", "cherry")
y=list(x)
y.append("mango")
x=tuple(y)
print(x)
```

```python
#Add tuple to a tuple
tuple1=("apple", "banana", "cherry")
tuple2=("mango",)
original_tuple=tuple1+tuple2
print(original_tuple)
```

```python
# Remove Items
x=("apple","banana","cherry")
y=list(x)
y.remove("apple")
x=tuple(y)
print(x)
```

```python
#Delete tuple
thistuple = ("apple", "banana", "cherry")
del thistuple #If try to print then it will give an error
```

```python
# Unpacking a Tuple
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)
```

```python
# Using Asterisk*
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green,yellow,*red)=fruits
print(green)
print(yellow)
print(red)
```

```python
# Loop Through a Tuple
mytuple=tuple(("apple","banana","cherry"))
for x in mytuple:
    print(x)
```

```python
# Loop Through the Index Numbers
thistuple = ("apple", "banana", "cherry")
for i in range(len(thistuple)):
  print(thistuple[i])
```

```python
# Using a While Loop
thistuple = ("apple", "banana", "cherry")
i = 0
while i < len(thistuple):
  print(thistuple[i])
  i = i + 1
```

```python
# Join Two Tuples
tuple1 = ("a", "b" , "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```

```python
# Multiply Tuples
fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2

print(mytuple)
```
